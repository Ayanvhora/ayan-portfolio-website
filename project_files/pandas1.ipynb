```python
import pandas as pd
import numpy as np
df=pd.read_csv(r"C:\Users\ayanv\Downloads\messy_HR_data.csv")
print(df.head(5))
```

```python
df.columns = df.columns.str.strip()
df = df.applymap(lambda x: x.strip() if isinstance(x, str) else x)
print(df.head(3))
```

```python
df['Salary'] = df['Salary'].replace('SIXTY THOUSAND', '60000')
df['Salary'] = pd.to_numeric(df['Salary'], errors='coerce')
df['Salary'] = df['Salary'].fillna(0)
print(df['Salary'].head(20))  # Top 20 rows print thase
```

```python
df['Age'] = df['Age'].replace('thirty', '30')  # 'thirty' ને 30 માં ફેરવો
df['Age'] = pd.to_numeric(df['Age'], errors='coerce')  # string values ને number માં convert કરો
df['Age']=df['Age'].fillna(0)
print(df['Age'])
```

```python
# Step 4: Clean 'Joining Date' column
df['Joining Date'] = pd.to_datetime(df['Joining Date'], errors='coerce', dayfirst=True)
df['Joining Date'] = df['Joining Date'].fillna(pd.Timestamp("1900-01-01"))
print(df['Joining Date'])
```
